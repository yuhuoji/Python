```python
"""
PIO5
"""
import sys

lines = [line.strip() for line in sys.stdin.readlines() if line.strip()]
for line in lines:
    a, b = map(int, line.split())
    if a == 0 and b == 0:
        break
    print(a + b)
```

```python
"""
PIO7
"""
import sys

lines = [line.strip() for line in sys.stdin.readlines() if line.strip()]
ptr = 0
t = int(lines[ptr])
ptr += 1
for _ in range(t):
    n = int(lines[ptr])
    ptr += 1
    nums = map(int, lines[ptr].split())
    ptr += 1
    print(sum(nums))
```

```python
"""
PI8
"""
import sys

lines = [line.strip() for line in sys.stdin.readlines() if line.strip()]
ptr = 0
n, m = map(int, lines[ptr].split())
ptr += 1
ans = 0
for _ in range(n):
    nums = map(int, lines[ptr].split())
    ptr += 1
    ans += sum(nums)
print(ans)
```
