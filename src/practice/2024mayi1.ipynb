```python
import sys


def solve():
    # 快速读取所有输入数据
    input_data = sys.stdin.read().split()
    if not input_data:
        return

    n = int(input_data[0])
    a = [int(x) for x in input_data[1:]]

    total_sum = sum(a)
    first_mod2 = a[0] % 2

    # 1. 检查奇偶性一致性：所有 ai mod 2 必须相同
    for x in a:
        if x % 2 != first_mod2:
            print("NO")
            return

    # 2. 对总和 S 进行质因数分解 (时间复杂度 O(sqrt(S)))
    def get_prime_factors(m):
        factors = set()
        temp = m
        d = 2
        while d * d <= temp:
            if temp % d == 0:
                factors.add(d)
                while temp % d == 0:
                    temp //= d
            d += 1
        if temp > 1:
            factors.add(temp)
        return factors

    prime_candidates = get_prime_factors(total_sum)
    results = []

    # 3. 验证每一个质因数作为 GCD 的可行性
    for g in prime_candidates:
        if g == 2:
            # 只有初始全为偶数，GCD 才能是 2
            if first_mod2 == 0:
                results.append(2)
        else:
            # g 是奇质数的情况
            if first_mod2 == 1:
                # 初始为奇数，最终每个 ai 最小为 g (1*g)
                if total_sum >= n * g:
                    results.append(g)
            else:
                # 初始为偶数，最终每个 ai 最小为 2g (2*g 是偶数)
                if total_sum >= n * (2 * g):
                    results.append(g)

    # 4. 格式化输出
    if not results:
        print("NO")
    else:
        print("YES")
        results.sort()
        print(*(results))


if __name__ == "__main__":
    solve()
```
